# Stage 8.9 -- Inferno Feature Quality

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
BASE = Path('../data/gold/validation/map_feature_quality')
def load(name):
    return pd.read_parquet(BASE / f'{name}.parquet')
```

## Final Scorecard

```python
scorecard = load('map_quality_scorecard')
display(scorecard)
scorecard.plot.bar(x='category', y=['passed', 'warnings', 'failed'], stacked=True, figsize=(10, 4))
plt.tight_layout()
```

## Feature Missingness Distribution

```python
missing = load('map_feature_missingness')
display(missing.sort_values('missing_share', ascending=False).head(20))
missing.sort_values('missing_share', ascending=False).head(20).plot.bar(x='feature_name', y='missing_share', figsize=(12, 4))
plt.tight_layout()
```

## Zero Share Distribution

```python
profile = load('map_feature_quality_profile')
display(profile.sort_values('zero_share', ascending=False).head(20))
profile.sort_values('zero_share', ascending=False).head(20).plot.bar(x='feature_name', y='zero_share', figsize=(12, 4))
plt.tight_layout()
```

## Constant / Near-Constant Features

```python
display(profile[(profile['constant']) | (profile['near_constant']) | (profile['all_zero'])].head(50))
```

## Semantic Signal Health

```python
semantic = load('map_semantic_signal_health')
display(semantic)
semantic.plot.bar(x='semantic_id', y='round_signal_share', figsize=(8, 4))
plt.tight_layout()
```

## Feature Health by Demo

```python
demo = load('map_demo_quality_summary')
display(demo)
demo.plot.bar(x='parse_id', y=['degenerate_feature_count', 'semantic_failure_count'], figsize=(10, 4))
plt.tight_layout()
```

## T vs CT Summary

```python
side = load('map_side_feature_health')
display(side.sort_values('t_zero_share', ascending=False).head(20))
```
