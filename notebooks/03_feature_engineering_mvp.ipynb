```python
from pathlib import Path
import pandas as pd

BASE = Path('..')
feature_eligible_path = BASE / 'data/silver/parsed_demos/feature_eligible_demos.parquet'
round_features_path = BASE / 'data/gold/round_features/round_features_mvp.parquet'
region_presence_path = BASE / 'data/gold/region_presence/region_presence_by_round.parquet'
utility_events_path = BASE / 'data/gold/utility_events/utility_events.parquet'
feature_audit_path = BASE / 'data/gold/feature_audit/feature_audit.parquet'
```
