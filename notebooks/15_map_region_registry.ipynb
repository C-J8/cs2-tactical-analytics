Inspect Mirage registry metadata.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

BASE = Path('../data/gold/maps/map_registry')

def load_table(name):
    return pd.read_parquet(BASE / f'{name}.parquet')

registry = load_table('map_registry')
regions = load_table('map_region_registry')
semantic_groups = load_table('map_semantic_groups')
mapping = load_table('map_region_semantic_mapping')
bombsites = load_table('map_bombsite_registry')
coverage = load_table('map_feature_semantic_coverage')
unknowns = load_table('map_region_unknowns')
audit = load_table('map_registry_audit')
```

## Coordinate Sketch

```python
bbox = regions.dropna(subset=['x_min', 'x_max', 'y_min', 'y_max'])
if bbox.empty:
    print('No coordinate bounding boxes are registered yet; Mirage currently uses named Awpy/place areas.')
else:
    fig, ax = plt.subplots(figsize=(6, 6))
    for _, row in bbox.iterrows():
        rect = plt.Rectangle((row.x_min, row.y_min), row.x_max - row.x_min, row.y_max - row.y_min, fill=False)
        ax.add_patch(rect)
        ax.text(row.x_min, row.y_min, row.region_id)
    ax.set_aspect('equal', adjustable='box')
    plt.tight_layout()
```
