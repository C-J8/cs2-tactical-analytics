```python
import matplotlib.pyplot as plt
oof = pd.read_parquet(base / 'inferno_ab_oof_predictions.parquet')
cm = pd.crosstab(oof['true_label'], oof['predicted_label'])
fig, ax = plt.subplots(figsize=(4, 4))
ax.imshow(cm.reindex(index=['A','B'], columns=['A','B'], fill_value=0), cmap='Blues')
ax.set_xticks([0,1], ['A','B'])
ax.set_yticks([0,1], ['A','B'])
ax.set_xlabel('Predicted')
ax.set_ylabel('True')
ax.set_title('OOF confusion matrix')
for i in range(2):
    for j in range(2):
        ax.text(j, i, int(cm.reindex(index=['A','B'], columns=['A','B'], fill_value=0).iloc[i, j]), ha='center', va='center')
plt.tight_layout()
```

```python
null = pd.read_parquet(base / 'inferno_ab_null_permutation.parquet')
obs = pd.read_parquet(base / 'inferno_ab_oof_metrics.parquet').iloc[0]['macro_f1']
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(null['macro_f1'], bins=25, alpha=0.8)
ax.axvline(obs, color='crimson', linewidth=2)
ax.set_title('Exploratory null comparison')
ax.set_xlabel('Permutation macro F1')
plt.tight_layout()
```
