# Stage 4.1 Side Datasets and Round Progression

Validate side-specific feature datasets and progression context tables.

```python
from pathlib import Path
import pandas as pd

BASE = Path('..')
round_dir = BASE / 'data/gold/round_features'
progression_dir = BASE / 'data/gold/round_progression'
audit_dir = BASE / 'data/gold/feature_audit'
```

```python
t_side_all = pd.read_parquet(round_dir / 'round_features_t_side_all.parquet')
t_side_planted = pd.read_parquet(round_dir / 'round_features_t_side_planted.parquet')
ct_side = pd.read_parquet(round_dir / 'round_features_ct_side.parquet')
{'t_side_all': t_side_all.shape, 't_side_planted': t_side_planted.shape, 'ct_side': ct_side.shape}
```

```python
t_side_planted['target_site_model_label'].value_counts(dropna=False)
```

```python
t_side_all[t_side_all['target_site_model_label'].isna()][['round_feature_id', 'series_id', 'round_num', 'winner_side']].head()
```

```python
round_region_timeline = pd.read_parquet(progression_dir / 'round_region_timeline.parquet')
round_region_timeline.groupby(['window_type', 'window_start', 'window_end'])['round_feature_id'].nunique().reset_index().head(30)
```

```python
round_outcome_context = pd.read_parquet(progression_dir / 'round_outcome_context.parquet')
round_outcome_context[['round_feature_id', 'round_progression_signature', 'round_outcome_type', 'bomb_carrier_region_105_115']].head(10)
```

```python
death_context = pd.read_parquet(progression_dir / 'death_context_by_round.parquet')
death_context[death_context['is_first_death']]['death_region_group'].value_counts(dropna=False).head(20)
```

```python
bomb_carrier_timeline = pd.read_parquet(progression_dir / 'bomb_carrier_timeline.parquet')
no_plant_rounds = set(t_side_all[t_side_all['target_site_model_label'].isna()]['round_feature_id'])
bomb_carrier_timeline[bomb_carrier_timeline['round_feature_id'].isin(no_plant_rounds)][['round_feature_id', 'window_type', 'window_start', 'window_end', 'bomb_carrier_player', 'bomb_carrier_region_group']].head(20)
```
