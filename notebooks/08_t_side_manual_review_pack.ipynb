# Stage 5.2 -- T-side Manual Review Pack

Qualitative review bridge between ranked tactical findings and the first leakage-controlled A/B baseline.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DATA = ROOT / 'data/gold/analysis/t_side_manual_review'

def load(name):
    return pd.read_parquet(DATA / f'{name}.parquet')

review = load('manual_review_rounds')
evidence = load('manual_review_evidence_by_round')
summary = load('manual_review_summary')
decisions = load('manual_review_decision_template')
readiness = load('manual_review_model_readiness')
```

## Review summary

```python
display(summary.T.rename(columns={0: 'value'}))
review['t_round_outcome'].value_counts().plot(kind='bar', title='Review rounds by outcome')
plt.ylabel('Selected finding-round pairs')
plt.show()
```

## Priority queue

```python
priority_columns = ['review_round_id', 'finding_category', 'review_priority', 'opponent', 'round_num', 't_round_outcome', 'suggested_focus']
display(review[priority_columns].head(20))
```

## Plant A, plant B, and no-plant examples

```python
for outcome in ['plant_A', 'plant_B', 'no_plant']:
    print(f'\n{outcome}')
    display(review.loc[review['t_round_outcome'].eq(outcome), priority_columns].head(10))
```
