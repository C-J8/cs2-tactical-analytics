# Stage 8.0 -- Feature Contract & Freeze

Inspect the frozen feature contract.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

BASE = Path('../data/gold/features/feature_contract')

def load_table(name):
    return pd.read_parquet(BASE / f'{name}.parquet')

contract = load_table('feature_contract')
summary = load_table('feature_contract_summary')
map_readiness = load_table('feature_contract_map_readiness')
modeling = load_table('feature_contract_modeling_readiness')
dashboard = load_table('feature_contract_dashboard_readiness')
unknowns = load_table('feature_contract_unknowns')
audit = load_table('feature_contract_audit')
```

## Total Features

```python
display(audit)
print(f"total features: {len(contract)}")
```

## Feature Families

```python
family_counts = contract['feature_family'].value_counts().sort_values()
display(family_counts.rename_axis('feature_family').reset_index(name='features'))
family_counts.plot.barh(figsize=(8, 6), title='Feature families')
plt.tight_layout()
```
