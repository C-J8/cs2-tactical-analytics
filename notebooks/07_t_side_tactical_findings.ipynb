# Stage 5.1 T-side Tactical Findings

Rank conservative tactical candidates from the Stage 5 Gold EDA outputs.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

BASE = Path('..')
FINDINGS_DIR = BASE / 'data/gold/analysis/t_side_tactical_findings'
def load_table(name):
    return pd.read_parquet(FINDINGS_DIR / f'{name}.parquet')
```

```python
key_findings = load_table('t_side_key_findings')
region_diff = load_table('t_side_ab_region_differences')
utility_diff = load_table('t_side_ab_utility_differences')
timing = load_table('t_side_ab_timing_breakpoints')
no_plant = load_table('t_side_no_plant_failure_findings')
bomb = load_table('t_side_bomb_carrier_findings')
opponents = load_table('t_side_opponent_tendencies')
progression = load_table('t_side_progression_findings')
review_queue = load_table('t_side_manual_review_queue')
audit = load_table('t_side_findings_audit')
key_findings.head(20)
```

## Timing breakpoints

```python
interval_timing = timing[timing['window_type'] == 'interval'].sort_values('window_start')
interval_timing.plot(x='window_start', y=['region_signal_score', 'utility_signal_score', 'combined_signal_score'], marker='o', figsize=(10, 4))
plt.title('A/B tactical signal by interval window')
plt.xlabel('Window start (seconds)')
plt.ylabel('Audit score')
plt.tight_layout()
```
