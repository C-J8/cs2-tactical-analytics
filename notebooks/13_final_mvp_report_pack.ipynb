Inspect the consolidated final MVP report outputs.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

BASE = Path('../data/gold/reporting/final_mvp')
CANDIDATE_BASE = Path('../data/gold/modeling/t_side_ab_candidate')

def load_table(name):
    return pd.read_parquet(BASE / f'{name}.parquet')

project_summary = load_table('final_project_summary')
dataset_snapshot = load_table('final_dataset_snapshot')
stage_status = load_table('final_pipeline_stage_status')
tactical_findings = load_table('final_tactical_findings_summary')
candidate_summary = load_table('final_model_candidate_summary')
model_error_summary = load_table('final_model_error_summary')
limitations = load_table('final_limitations')
next_steps = load_table('final_next_steps')
artifact_manifest = load_table('final_artifact_manifest')
report_audit = load_table('final_report_audit')
confusion = pd.read_parquet(CANDIDATE_BASE / 'candidate_model_confusion_matrix.parquet')
```

## Confusion Matrix

```python
matrix = confusion.pivot(index='true_label', columns='predicted_label', values='count').fillna(0)
display(matrix)
ax = matrix.plot(kind='bar', figsize=(6, 4), rot=0)
ax.set_title('Candidate confusion matrix')
ax.set_xlabel('True label')
ax.set_ylabel('Rounds')
plt.tight_layout()
```
