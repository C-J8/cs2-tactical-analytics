# Stage 8.7 -- Inferno Region Mapping

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
BASE = Path('../data/gold/maps/inferno/region_mapping')
AREA = Path('../data/gold/maps/area_discovery')
def load(name):
    return pd.read_parquet(BASE / f'{name}.parquet')
proposal = load('inferno_region_mapping_proposal')
crosswalk = load('inferno_place_region_crosswalk')
physical = load('inferno_physical_region_inventory')
semantic = load('inferno_semantic_mapping')
coverage = load('inferno_semantic_coverage')
coordinate = load('inferno_region_coordinate_validation')
candidate = load('inferno_candidate_feature_portability_v2')
unknowns = load('inferno_region_mapping_unknowns')
audit = load('inferno_region_mapping_audit')
sample = pd.read_parquet(AREA / 'map_place_coordinate_sample.parquet')
sample = sample[sample['map_id'].eq('inferno')]
```
