```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
BASE = Path('../data/gold/analysis/multi_map_tactical_eda')
def load(name):
    return pd.read_parquet(BASE / f'{name}.parquet')
```

```python
scope = load('multi_map_eda_scope_inventory')
display(scope)
scope.plot.bar(x='map_id', y=['t_rounds','planted_t_rounds','no_plant_t_rounds'], figsize=(8,4))
plt.tight_layout()
```
