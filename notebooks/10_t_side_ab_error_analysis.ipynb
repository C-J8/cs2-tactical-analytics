# Stage 6.1 -- T-side A/B Baseline Error Analysis

Interpretation of existing out-of-fold baseline predictions without retraining.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DATA = ROOT / 'data/gold/modeling/t_side_ab_error_analysis'

def load(name):
    return pd.read_parquet(DATA / f'{name}.parquet')

overview = load('ab_error_overview')
by_model = load('ab_error_by_horizon_model')
errors = load('ab_error_rounds')
high = load('ab_high_confidence_errors')
by_opponent = load('ab_error_by_opponent')
by_label = load('ab_error_by_true_label')
stability = load('ab_feature_importance_stability')
contrast = load('ab_feature_error_contrast')
horizons = load('ab_horizon_practical_recommendation')
queue = load('ab_error_manual_review_queue')
```

## Overview and horizon metrics

```python
display(overview.T.rename(columns={0: 'value'}))
display(by_model)
by_model.plot(x='horizon_seconds', y=['macro_f1', 'majority_macro_f1_same_horizon'], marker='o', title='Selected model vs majority baseline')
plt.ylabel('Macro F1')
plt.show()
```

## Errors by class and high-confidence errors

```python
display(by_label)
display(high[['horizon_seconds', 'model_name', 'opponent', 'round_num', 'true_label', 'predicted_label', 'prediction_confidence', 'suggested_error_reason']].head(20))
errors['error_type'].value_counts().plot(kind='bar', title='Selected-model error directions')
plt.ylabel('Errors')
plt.show()
```
